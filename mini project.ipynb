```python
import pandas as pd 
df = pd.read_csv("student_performance_eda_project.csv")
print(df)
```

```python
# Part A project task Data understanding and cleaning
# 2nd Answer
print(df.head(10))
print(df.shape)
print(df.columns)
print(df.dtypes)
```

```python
#3rd Answer
print(df.isnull().sum())
print("\nDuplicated records:", df.duplicated().sum())
```

```python
#4th Answer
print(df.describe(include='all'))
print(df[df['Final_Exam_Marks_Out_Of_100']>100])
print(df[df['Final_Exam_Marks_Out_Of_100']<0])

#5th Answer
print(df.describe())
```

```python
#6th Answer Numerical and categorical
num_col = df.select_dtypes(include='number').columns.tolist()
print(num_col)
cat_col = df.select_dtypes(include='str').columns.tolist()
print(cat_col)
```

```python
#Part B 
#Univariate analysis
#7th Answer
print(df.columns.tolist())  #took this cell to know columns names
print(df['Final_Exam_Marks_Out_Of_100'].mean())
print(df['Final_Exam_Marks_Out_Of_100'].median())
print(df['Final_Exam_Marks_Out_Of_100'].std())
```

```python
#8th Answer
import matplotlib.pyplot as plt
plt.hist(df['Final_Exam_Marks_Out_Of_100'],bins = 15)
plt.title('Histogram of final exam marks')
plt.xlabel('Final Exam Marks Out Of 100')
plt.ylabel('Number of students')
plt.show()
```

```python
# 9th Answer count plot
dept_count = df['Department'].value_counts()
plt.bar(dept_count.index , dept_count.values)
plt.title('Number of Students in Each Department')
plt.xlabel('Department')
plt.ylabel('Count')
plt.show()
```

```python
#10 Answer
plt.boxplot(df['Study_Hours_Per_Day'].dropna())
plt.title('Boxplot of study hours')
plt.ylabel('study Hours Per Day')
plt.show()

plt.boxplot(df['Final_Exam_Marks_Out_Of_100'].dropna())
plt.title('Boxplot of final exam')
plt.ylabel('Final Marks')
plt.show()
```

```python
#Part C Bivariate ScatterPlot
plt.scatter(df['Study_Hours_Per_Day'],df['Final_Exam_Marks_Out_Of_100'])
plt.title('Study Hours vs Final exam marks')
plt.xlabel('Study hours per day')
plt.ylabel('Final Exam Marks Out of 100')
plt.show()
```

```python
#12 Answer
dept_avg = df.groupby('Department')['Final_Exam_Marks_Out_Of_100'].mean()
plt.bar(dept_avg.index , dept_avg.values)
plt.title('Average Final marks across dept')
plt.xticks(rotation = 45)
plt.ylabel('Average marks')
plt.show()
print(dept_avg)
```

```python
#13/ 
tutor_avg = df.groupby('Private_Tutoring')['Final_Exam_Marks_Out_Of_100'].mean()
plt.bar(tutor_avg.index.astype(str),tutor_avg.values)
plt.title('Private Tutoring vs Final Marks(Average)')
plt.xlabel('Private Tutoring')
plt.ylabel('Average Final Marks')
plt.show()
```

```python
#14/
corr = df.select_dtypes(include=['int64','float64']).corr()
print(corr)
```

```python
#15 /
plt.imshow(corr, cmap='coolwarm',interpolation='nearest')
plt.colorbar()
plt.xticks(range(len(corr.columns)) , corr.columns,rotation = 45)
plt.yticks(range(len(corr.columns)) , corr.columns)
plt.title('Correlation Matrix Heatmap')
plt.show()
```

```python
#16 /
plt.scatter(df['Attendance_Percentage'] , df['Final_Exam_Marks_Out_Of_100'])
plt.title('Attendance vs Final Marks')
plt.xlabel('Attendance')
plt.ylabel('Fianl Marks')
plt.show()
```
